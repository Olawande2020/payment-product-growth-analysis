```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Customer Segmentation

```python
transactions = pd.read_csv("transactions.csv")
transactions.head()
```

```python
user_revenue = transactions.groupby("user_id")["amount"].sum().reset_index()
user_revenue
```

```python
user_revenue["segment"] = pd.qcut(
    user_revenue["amount"],
    q=3,
    labels=["Low Value", "Mid Value", "High Value"]
)
```

```python
user_revenue.segment.value_counts()
```

### Revenue is highly concentrated among a small group of high-value users, indicating an opportunity to prioritize retention and engagement strategies for this segment.
